```python
import numpy as np

total_spent = cleaned_df['Total Spent'].dropna().to_numpy(dtype=float)

print("Transactions with total spent:", total_spent.size)
print("Average amount spent:", round(np.mean(total_spent), 2))
print("Median amount spent:", round(np.median(total_spent), 2))
print("Smallest amount:", np.min(total_spent))
print("Largest amount:", np.max(total_spent))
```
